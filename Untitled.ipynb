```python
spec = mj.MjSpec.from_file(piper_mj_description.MJCF_PATH)
# spec = mj.MjSpec.from_file(MJ_MODEL_PATH.as_posix())
model = spec.compile()
data = mj.MjData(model)
joint_indices = [model.joint(name).id for name in JOINT_NAMES]
qpos_indices = model.jnt_qposadr[joint_indices]
qvel_indices = model.jnt_dofadr[joint_indices]
qpos_range = model.jnt_range[joint_indices]

def random_qpos():
    return np.random.uniform(qpos_range[:, 0], qpos_range[:, 1])

def tau_gravity(joint_positions):
    assert len(joint_positions) == len(JOINT_NAMES)
    data.qpos[qpos_indices] = joint_positions
    mj.mj_forward(model, data)
    return data.qfrc_bias[qvel_indices]

def mocked_tau_gravity(torque, noise=False):
    gravity_torques = 0.1 * np.power(torque, 3) + 4 * np.power(torque, 2) + 2 * torque + 10
    if noise:
        gravity_torques += np.random.normal(0.0, 0.05, len(gravity_torques))
    return gravity_torques
```

# Using cubic approximator

```python
def cubic_gravity_tau(sim_torque, a, b, c, d):
  """A cubic adjustment of sim-predicted gravity torques."""
  return (
      a * sim_torque * sim_torque * sim_torque +
      b * sim_torque * sim_torque +
      c * sim_torque +
      d
  )
```

# Using GP

```python
import numpy as np
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, Matern, ConstantKernel as C
from scipy.optimize import minimize
from scipy.optimize import differential_evolution
import matplotlib.pyplot as plt

class ArmDynamicsGP:
    def __init__(self):
        """
        Gaussian Process for learning arm dynamics (gravity torques).
        """
        self.X_observed = []  # Joint configurations we've measured
        self.y_observed = []  # Gravity torques we've measured
        
        # Define GP kernel - you might want to customize this based on arm kinematics
        # RBF kernel assumes smooth function, Matern is more flexible
        kernel = C(1.0, (1e-3, 1e3)) * RBF(length_scale=1.0, length_scale_bounds=(1e-2, 1e2))
        # from sklearn.gaussian_process.kernels import WhiteKernel
        # kernel = (C(1.0, (1e-3, 1e3)) * 
        #           Matern(length_scale=1.0, length_scale_bounds=(1e-2, 1e2), nu=2.5) + 
        #           WhiteKernel(noise_level=0.01))

        self.gp = GaussianProcessRegressor(
            kernel=kernel,
            n_restarts_optimizer=10,
            alpha=1e-6,  # Small noise term for numerical stability
            normalize_y=True
        )
        
    def add_observation(self, mj_gravity_torques, actual_gravity_torques):
        """Add a new observation to our dataset."""
        self.X_observed.append(mj_gravity_torques)
        self.y_observed.append(actual_gravity_torques)
        
    def fit(self):
        """Fit the GP to all observations so far."""
        if len(self.X_observed) < 2:
            raise ValueError("Need at least 2 observations to fit GP")
            
        X = np.array(self.X_observed)
        y = np.array(self.y_observed)
        # from sklearn.preprocessing import StandardScaler
        # X_scaler = StandardScaler()
        # y_scaler = StandardScaler()

        # X_scaled = X_scaler.fit_transform(X)
        # y_scaled = y_scaler.fit_transform(y)
                
        self.gp.fit(X, y)
        # self.gp.fit(X_scaled, y_scaled)
        
    def predict(self, joint_configs, return_std=True):
        """Make predictions at given joint configurations."""
        return self.gp.predict(tau_gravity(joint_configs).reshape(1, -1), return_std=return_std)
        
    def acquisition_function(self, joint_config):
        """
        Acquisition function: Maximum Posterior Variance
        
        Returns the negative variance (since we'll minimize this function
        to find the maximum variance point).
        """
        # print(joint_config)
        _, std = self.predict(joint_config, return_std=True)
        
        # Return negative variance (negative because scipy.optimize minimizes)
        return -np.sum(std**2)
    
    def find_next_sample_point(self, bounds):
        """
        Find the next point to sample using the acquisition function.
        
        Args:
            bounds: List of (min, max) tuples for each joint angle
            
        Returns:
            Next joint configuration to sample
        """
        # Random restart optimization to find global maximum of acquisition function
        best_acquisition = float('inf')
        best_point = None
        
        n_restarts = 20
        
        for _ in range(n_restarts):
            # Random starting point within bounds
            x0 = random_qpos()
            
            # Minimize negative acquisition (i.e., maximize acquisition)
            # result = minimize(
            #     self.acquisition_function,
            #     x0,
            #     bounds=bounds,
            #     # method='L-BFGS-B'
            #     method='L-BFGS-B'
            # )
            result = differential_evolution(
                self.acquisition_function,
                bounds,
                seed=42,
                workers=1,  # or -1 for parallel
                atol=1e-6
            )
            
            if result.success and result.fun < best_acquisition:
                best_acquisition = result.fun
                best_point = result.x
                
        return best_point


def bayesian_optimization_loop(arm_gp, n_iterations=20):
    """
    Main Bayesian Optimization loop for learning arm dynamics.
    
    Args:
        arm_gp: ArmDynamicsGP instance
        measurement_function: Function that takes joint_config and returns gravity_torques
        bounds: Joint angle bounds for each joint
        n_iterations: Number of BO iterations to run
    """
    
    # Step 1: Initialize with a few random points
    print("Initializing with random samples...")
    n_initial = 5
    
    for i in range(n_initial):
        # Sample random joint configuration
        # joint_config = random_qpos()
        
        # Get measurement
        gravity_torques = tau_gravity(random_qpos())
        mocked_gravity_torques = mocked_tau_gravity(gravity_torques)
        
        # Add to GP dataset
        arm_gp.add_observation(gravity_torques, mocked_gravity_torques)
        print(f"Initial sample {i+1}: joints={gravity_torques.round(3)}, torques={mocked_gravity_torques.round(3)}")
    
    # Step 2: Bayesian optimization loop
    print(f"Starting Bayesian optimization for {n_iterations} iterations...")
    
    for iteration in range(n_iterations):
        # Fit GP to current data
        arm_gp.fit()
        
        # Find next point with maximum acquisition function (max variance)
        next_point = arm_gp.find_next_sample_point(qpos_range)
        
        if next_point is None:
            print(f"Optimization failed at iteration {iteration}")
            break
            
        # Get measurement at chosen point
        gravity_torques = tau_gravity(next_point)
        mocked_gravity_torques = mocked_tau_gravity(gravity_torques) # TODO: Add noise
        # Add observation
        arm_gp.add_observation(gravity_torques, mocked_gravity_torques)
        
        # Report progress
        _, std = arm_gp.predict([next_point])
        print(f"Iteration {iteration+1}: joints={next_point.round(3)}, "
              f"torques={gravity_torques.round(3)}, uncertainty={std.round(3)}")


if __name__ == "__main__":
    # Create GP model
    arm_gp = ArmDynamicsGP()
    
    # Run Bayesian optimization
    bayesian_optimization_loop(
        arm_gp, 
        n_iterations=15
    )
    
    print(f"Final model trained on {len(arm_gp.X_observed)} samples")
    
    # Test the trained model
    test_config = random_qpos()
    pred_mean, pred_std = arm_gp.predict([test_config])
    true_torques = tau_gravity(test_config)
    
    print(f"Test at joints={test_config}:")
    print(f"Predicted torques: {pred_mean[0].round(3)} ± {pred_std.round(3)}")
    print(f"True torques: {true_torques.round(3)}")
```
